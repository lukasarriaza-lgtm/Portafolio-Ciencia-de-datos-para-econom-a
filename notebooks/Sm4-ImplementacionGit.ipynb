## Parte 1: Autenticación Segura en Computadores de Laboratorio

Para evitar dejar tus credenciales guardadas en computadores compartidos, usamos la herramienta **GitHub CLI (`gh`)**.

### Paso de Seguridad: Verificar y Cerrar Sesión Anterior
Antes de iniciar tu sesión, comprueba si el estudiante anterior dejó su cuenta abierta en el computador:
```bash
gh auth status
```
Si la terminal indica que hay una sesión activa de otra persona, ciérrala ejecutando:
```bash
gh auth logout
```
*(Presiona **`y`** para confirmar la desvinculación).* 

---

### Iniciar tu Sesión (Paso a paso en la Terminal):
```bash
gh auth login
```

1. `Where do you use GitHub?` $\rightarrow$ **`GitHub.com`**
2. `What is your preferred protocol?` $\rightarrow$ **`HTTPS`** (o **`SSH`**)
3. `Authenticate Git with your GitHub credentials?` $\rightarrow$ **`Y`** (Yes)
4. `How would you like to authenticate?` $\rightarrow$ **`Login with a web browser`**
5. Copia el **código de 8 caracteres** que te muestra la consola, presiona **Enter**, pégalo en la ventana web del navegador y autoriza.

### Celda de Verificación de Autenticación:
Ejecuta esta celda para comprobar si tu terminal ya está autenticada con tu cuenta:

import subprocess
res = subprocess.run(["gh", "auth", "status"], capture_output=True, text=True)
print(res.stdout or res.stderr)

## Parte 2: Vinculación del Proyecto con GitHub

### Reiniciar o Limpiar un Git Previo (Desinicializar)
Si la carpeta ya contenía un Git desconfigurado o si otro estudiante inició Git en la carpeta por error, ejecuta en la **Terminal** dentro de la carpeta del proyecto para empezar 100% limpio desde cero:

```bash
rm -rf .git
```
*(Esto borra la carpeta oculta `.git` y deja el proyecto limpio sin alterar tus archivos de código o cuadernos).* 

---

### Vinculación Inicial a GitHub:

**A) Si es la primera vez que vinculas el repositorio (Nuevo):**
```bash
git init
git remote add origin git@github.com:nombre_usuario/nombre_repositorio.git
git branch -M main
git add .
git commit -m "Primer commit oficial del proyecto"
git push -u origin main --force
```

**B) Si ya existía un remoto `origin` anterior y quieres cambiar la URL:**
```bash
git remote set-url origin git@github.com:nombre_usuario/nombre_repositorio.git
git push -u origin main --force
```

## Parte 4: Trabajo Colaborativo con Ramas (Branches), Pull Requests y Merge

En proyectos de equipo, nadie trabaja directo sobre `main`. Cada estudiante trabaja en su propia rama:

### 4.1 Crear y cambiarse a una rama nueva:
```bash
git switch -c feat/analisis-inflacion
```

### 4.2 Subir la rama individual a GitHub:
Recordatorio importante: debes indicar el remoto `origin` antes del nombre de la rama:
```bash
git push -u origin feat/analisis-inflacion
```

---
### 4.3 Formas de Integrar (Merge) los Cambios de la Rama a `main`:

#### 🔹 Método A: Vía Página Web de GitHub (Pull Request / PR - Recomendado para Equipos)
1. **Crear el Pull Request**: Entra al enlace que te muestra la terminal o ve a tu repositorio en GitHub y haz clic en el botón amarillo **`Compare & pull request`**.
2. **Revisar y Fusionar (Merge)**: Revisa los cambios y presiona **`Merge pull request`** $\rightarrow$ **`Confirm merge`**. Esto integra la rama a `main` en GitHub.
3. **Sincronizar localmente**: Vuelve a `main` en tu terminal y descarga los cambios:
   ```bash
   git switch main
   git pull origin main
   ```

#### 🔹 Método B: Vía Terminal Local con `git merge` (Rápido / Trabajo Individual)
Si prefieres unir las ramas directamente en tu computador sin usar la web:
1. **Cambiar a `main`**: 
   ```bash
   git switch main
   ```
2. **Fusionar la rama localmente**:
   ```bash
   git merge feat/analisis-inflacion
   ```
3. **Subir el `main` actualizado a GitHub**:
   ```bash
   git push origin main
   ```

---
### 4.4 Revisar si hay modificaciones en GitHub (`git fetch` vs `git pull` vs `git clone`):

1. **Revisar si hay cambios en GitHub sin modificar tu código local:**
   ```bash
   git fetch
   git status
   ```
   *(Si hay commits nuevos en GitHub, `git status` dirá: `Your branch is behind 'origin/main' by X commits`).*

2. **Descargar e integrar los cambios en tu equipo:**
   ```bash
   git pull origin main
   ```

> **Diferencia Clave:**
> * `git clone`: Se usa **una sola vez** en un computador nuevo para descargar el proyecto desde cero por primera vez.
> * `git fetch`: Inspecciona si hay actualizaciones en GitHub sin alterar tus archivos locales.
> * `git pull`: Descarga e integra los avances de tus compañeros en tu carpeta local.

---
### 4.5 Caso Práctico: Trabajar en Ramas Paralelas para Tareas Independientes

¿Qué pasa si dos estudiantes trabajan al mismo tiempo en tareas independientes que no dependen la una de la otra? (Ej. Estudiante A hace `feat/tarea-a` y Estudiante B hace `feat/tarea-b`):

1. **Estudiante A sube y fusiona primero la Tarea A:**
   ```bash
   git push -u origin feat/tarea-a
   ```
   *(Abre el Pull Request en GitHub y hace Merge. Ahora `main` en GitHub tiene la Tarea A).* 

2. **Estudiante B sube y fusiona después la Tarea B:**
   ```bash
   git push -u origin feat/tarea-b
   ```
   *(Abre el Pull Request en GitHub y hace Merge. Como la Tarea B es independiente, GitHub une automáticamente la Tarea B al lado de la Tarea A sin ningún conflicto).* 

3. **Sincronización Final de Todo el Equipo:**
   Para que todos los integrantes tengan el proyecto completo con las **dos tareas unificadas**, cada estudiante ejecuta en su terminal:
   ```bash
   git switch main
   git pull origin main
   ```

## Parte 5: Resolución de Conflictos (Merge Conflicts)

Si dos integrantes modificaron la misma línea del mismo archivo, Git mostrará marcas en el código:

```python
<<<<<<< HEAD  (Tu versión)
df = pd.read_csv("data/inflacion.csv")
=======
df = pd.read_csv("data/coyuntura.csv")
>>>>>>> feat/modelo-sql  (Versión del compañero)
```

**Pasos para resolver:**
1. En **VS Code**, usa los botones flotantes sobre el conflicto (`Accept Current`, `Accept Incoming` o `Accept Both`).
2. Guarda el archivo e integra los cambios:
   ```bash
   git add .
   git commit -m "Resuelve conflicto de fusión"
   git push
   ```

## Parte 6: Control de Daños y Deshacer Errores

| Situación | Comando exacto | ¿Qué hace? |
| :--- | :--- | :--- |
| **Descartar cambios locales no guardados** | `git restore <archivo>` | Vuelve el archivo a su estado original guardado |
| **Sacar archivo del staging (`git add`)** | `git restore --staged <archivo>` | Deshace el `git add` del archivo |
| **Deshacer commit público de forma segura** | `git revert <hash_commit>` | Crea un nuevo commit que deshace el error sin borrar el historial |
| **Deshacer último commit local (conservando código)** | `git reset --soft HEAD~1` | Deshace el commit pero guarda tus cambios en el disco |
| **Deshacer último commit local (destructivo)** | `git reset --hard HEAD~1` | ⚠️ Elimina el commit y borra los cambios realizados |

## Parte 7: El Archivo `.gitignore` (Seguridad de Credenciales y Datos)

El archivo `.gitignore` le indica a Git qué archivos o carpetas **NUNCA** deben ser rastreados ni subidos a GitHub.

### La Regla Fundamental:
El archivo `.env` contiene tus claves de API (Banco Central de Chile, FRED, etc.). **Nunca** se sube a GitHub. Un repositorio es público y permanente: si subes una clave por error, queda en el historial de GitHub aunque después borres el archivo.

### Ejemplo de un `.gitignore` típico del proyecto:
```gitignore
# Credenciales de API - NUNCA SUBIR
.env

# Bases de datos locales y descargas masivas
data/*.db
data/*.csv
!data/.gitkeep

# Archivos temporales de Python y Jupyter
__pycache__/
*.pyc
.ipynb_checkpoints/
venv/
.venv/
```

### Celda de Verificación del `.gitignore`:
Ejecuta esta celda para comprobar que el archivo `.gitignore` existe en la raíz y excluye `.env`:

````python
from pathlib import Path

gitignore = Path("../.gitignore")
if not gitignore.exists():
    gitignore = Path(".gitignore")

if gitignore.exists():
    contenido = gitignore.read_text(encoding="utf-8")
    print("Contenido del .gitignore:")
    print("-" * 40)
    print(contenido.strip())
    print("-" * 40)
    print("Excluye .env correctamente:", ".env" in contenido)
else:
    print("⚠️ No se encontró el archivo .gitignore. Crea uno antes de hacer commits.")
````

## Parte 8: Celda de Diagnóstico del Repositorio Local

Ejecuta esta celda para verificar la URL remota y el estado actual de tus archivos:

````python
import subprocess

print("=== Remoto Configurado ===")
remoto = subprocess.run(["git", "remote", "-v"], capture_output=True, text=True)
print(remoto.stdout if remoto.stdout else "Sin remoto configurado")

print("=== Estado Actual ===")
estado = subprocess.run(["git", "status"], capture_output=True, text=True)
print(estado.stdout)
````
